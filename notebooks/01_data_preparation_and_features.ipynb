```python
AUDIO_DIR = Path("../data/raw/large_dataset_3")
OUTPUT_CSV = Path("../data/new_features_3sec.csv")

if AUDIO_DIR.exists():
    table = feature_table_from_directory(AUDIO_DIR)
    OUTPUT_CSV.parent.mkdir(parents=True, exist_ok=True)
    table.to_csv(OUTPUT_CSV, index=False)
    print("Saved:", OUTPUT_CSV)
    print("Shape:", table.shape)
else:
    print("Raw large audio directory is not present.")
    print("Recovered historical output shape: 80,207 rows x 59 columns.")
```
