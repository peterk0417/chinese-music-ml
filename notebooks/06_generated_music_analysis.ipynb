# 05. Generated music analysis

Reconstructs the surviving source-vs-generated comparison using:

- waveform
- STFT
- mel-spectrogram
- MFCC

The original notebook used `1.1.wav` as the source and `pred_rnn2.wav` as the generated sample.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
```

```python
import numpy as np
import matplotlib.pyplot as plt
import librosa
import librosa.display

SOURCE = ROOT / "data" / "generation_sources" / "1.1.wav"
GENERATED = ROOT / "data" / "generated" / "pred_rnn2.wav"
```

```python
y_source, sr_source = librosa.load(SOURCE, mono=True)
y_generated, sr_generated = librosa.load(GENERATED, mono=True)
y_source = librosa.util.normalize(y_source)
y_generated = librosa.util.normalize(y_generated)
```

```python
fig, ax = plt.subplots(4, 2, figsize=(14, 12))

librosa.display.waveshow(y_source, sr=sr_source, ax=ax[0, 0])
ax[0, 0].set_title("Source waveform")
librosa.display.waveshow(y_generated, sr=sr_generated, ax=ax[0, 1])
ax[0, 1].set_title("Generated waveform")

for col, (y, sr, title) in enumerate([
    (y_source, sr_source, "Source"),
    (y_generated, sr_generated, "Generated"),
]):
    stft = librosa.amplitude_to_db(np.abs(librosa.stft(y)), ref=np.max)
    img = librosa.display.specshow(stft, sr=sr, x_axis="time", y_axis="hz", ax=ax[1, col])
    ax[1, col].set_title(f"{title} STFT")
    fig.colorbar(img, ax=ax[1, col], format="%+2.0f dB")

    mel = librosa.feature.melspectrogram(y=y, sr=sr)
    mel_db = librosa.power_to_db(mel, ref=np.max)
    img = librosa.display.specshow(mel_db, sr=sr, x_axis="time", y_axis="hz", ax=ax[2, col])
    ax[2, col].set_title(f"{title} mel-spectrogram")
    fig.colorbar(img, ax=ax[2, col], format="%+2.0f dB")

    mfcc = librosa.feature.mfcc(y=y, sr=sr)
    img = librosa.display.specshow(mfcc, x_axis="time", ax=ax[3, col])
    ax[3, col].set_title(f"{title} MFCC")
    fig.colorbar(img, ax=ax[3, col])

plt.tight_layout()
```
